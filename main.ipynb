```python
import re, sqlite3

def _looks_like_salary(raw):
    if not raw or raw in ("N/A", "NC", "Non communiqué", "Non spécifié"):
        return False
    has_number   = bool(re.search(r'\b\d{4,6}\b', raw))
    has_currency = any(k in raw.lower() for k in ["dh", "mad", "dirham"])
    return has_number and has_currency

conn = sqlite3.connect("jobs.db")
import pandas as pd
df = pd.read_sql("SELECT salary FROM jobs", conn)
conn.close()

passing = df[df["salary"].apply(_looks_like_salary)]["salary"]
print(f"Passing: {len(passing)}\n")
for s in passing:
    matched = [k for k in ["dh","mad","dirham"] if k in s.lower()]
    print(f"  currency={matched} | {s[:80]}")
```
